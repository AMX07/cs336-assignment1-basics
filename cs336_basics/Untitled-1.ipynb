```python
L = 48

vocab_size = 50257
# context_length = 1024
T =1024

# num_layers: 48
d = 1600
# num_heads= 25
h = 25
d_ff = 4288






eq1 = L * (2*d + 4*d**2 + 3*d*d_ff) + 2*vocab_size*d + d

eq2 = L * (2*d + 4*d**2 + 3*d*d_ff) + 2*vocab_size*d + d

eq3 = L * (4*d + 8*d**2 + 6*d*d_ff) + 4*vocab_size*d + 2*d

eq4 = (
    L * (8*T*d + 4*T*d_ff + 2*h*T**2)
    + 2*T*vocab_size
    + T*d
)

# Simplified total: equivalent to eq1 + eq2 + eq3 + eq4
total = (
    L * (
        16*d**2
        + 4*d_ff*(3*d + T)
        + 8*d*(T + 1)
        + 2*h*T**2
    )
    + 2*vocab_size*(4*d + T)
    + d*(T + 4)
)

print(total)
```
